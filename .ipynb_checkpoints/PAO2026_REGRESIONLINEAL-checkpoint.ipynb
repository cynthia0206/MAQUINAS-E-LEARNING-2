```python
#Importamos las librerias

import pandas as pd
from sklearn.linear_model import LinearRegression
```

```python
#Creamos el conunto de datos 

df = pd.DataFrame ({
    "horas":[1,2,3,4,5],
    "nota":[2,3,5,4,6]
})
X= df[["horas"]]
y= df[["nota"]]
print (X)
```

```python
#Creamos nuestro modelo
modelo = LinearRegression()
modelo.fit(X, y)
```

```python
#Imprimimos el interceptor y el coef
print("Interceptor: ", modelo.intercept_)
print("Pendiente:", modelo.coef_[0])
```

```python
#Valores predichos 
y_pred= modelo.predict(X)
```

```python
#REalizamos la nueva predicción 7
horas=float(input("Ingrese las horas de estudioa"))
prediccion = modelo.predict([[horas]])
print("Predicción:" , prediccion)
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Creamos los puntos de la recta
X_linea = np.linspace(1,6,100)
X_linea_df = pd.DataFrame({"horas": X_linea})
```

```python

# Realizamos los datos predichos
y_predichos = modelo.predict(X)

# Graficamos los datos reales
plt.scatter(X, y, label="Datos reales")

# Graficamos los datos predichos
plt.scatter(X, y_predichos, label="Datos predichos")

# Graficamos la recta de regresión
plt.plot(X_linea, y_linea, label="Recta de regresión")

# Graficamos la nueva predicción
plt.scatter(horas, prediccion[0][0], marker="X", s=100, label="Nueva predicción")

# Agregamos títulos y etiquetas
plt.title("Regresión Lineal")
plt.xlabel("Horas de estudio")
plt.ylabel("Nota")

plt.legend()
plt.show()
```
