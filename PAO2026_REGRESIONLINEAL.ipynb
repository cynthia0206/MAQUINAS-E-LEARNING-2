```python
#Importamos las librerias

import pandas as pd
from sklearn.linear_model import LinearRegression
```

```python
#Creamos el conunto de datos 

df = pd.DataFrame ({
    "horas":[1,2,3,4,5],
    "nota":[2,3,5,4,6]
})
X= df[["horas"]]
y= df[["nota"]]
print (X)
```

```python
#Creamos nuestro modelo
modelo = LinearRegression()
modelo.fit(X, y)
```

```python
#Imprimimos el interceptor y el coef
print("Interceptor: ", modelo.intercept_)
print("Pendiente:", modelo.coef_[0])
```

```python
#REalizamos la nueva predicción
prediccion = modelo.predict([[6]])
print("Predicción:" , prediccion)
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Creamos los puntos de la recta
X_linea = np.linspace(1,6,100)
X_linea_df = pd.DataFrame({"horas": X_linea})
```

```python
# Realizamos las predicciones para la recta
y_linea = modelo.predict(X_linea_df)

# Graficamos los datos originales
plt.scatter(X, y, label="Datos reales")

# Graficamos la recta de regresión
plt.plot(X_linea, y_linea, label="Recta de regresión")

# Graficamos la nueva predicción
plt.scatter(6, prediccion[0][0], marker="X", s=100, label="Nueva predicción")

# Agregamos títulos y etiquetas
plt.title("Regresión Lineal")
plt.xlabel("Horas de estudio")
plt.ylabel("Nota")

# Mostramos la leyenda
plt.legend()

# Mostramos la gráfica
plt.show()
```
