
## Construir redes

Las capas de una red neuronal deben ser compatibles en sus dimensiones de entrada (input) y salida (output). La salida de una capa debe coincidir con la entrada que espera la siguiente capa.



## Tipos de capas

Cada tipo de capa tiene un uso dependiendo del tipo de datos y del problema que se desea resolver.

- **Dense:** procesa datos estructurados y vectores.
- **LSTM:** procesa secuencias de datos, como series temporales.
- **Convolucionales (CNN):** procesan imágenes y extraen características visuales.



## Estructura de la red neuronal

- La estructura más habitual es secuencial (feedforward).
- Existen otras arquitecturas, como Two-branch, Multihead e Inception.
- La arquitectura ideal depende del problema, la experiencia y la experimentación.



## Deep Learning en Computer Vision

- Redes neuronales convolucionales (CNN).
- Entrenamiento de redes neuronales.
- Trabajo con pocos datos.
- Data augmentation.
- Transfer Learning y Fine-tuning.
- Visualización de lo que aprenden las CNN.



## Convolución

- Las capas Dense pueden aprender patrones globales de los datos.
- Las convoluciones detectan patrones locales en las imágenes.
- Estos patrones pueden ser líneas, curvas, bordes y formas.



Las convoluciones permiten aprender patrones de manera progresiva. Las primeras capas pueden reconocer líneas y bordes; las capas posteriores combinan esas características para reconocer formas más complejas.
